```python
import os
import json
from pathlib import Path

# Diretório do projeto
projeto_path = Path("../..")

# 1. Contar arquivos por tipo
def contar_arquivos(extensao):
    count = 0
    for root, dirs, files in os.walk(projeto_path):
        # Ignora .venv e .git
        dirs[:] = [d for d in dirs if d not in ['.venv', '.git', '__pycache__']]
        for file in files:
            if file.endswith(extensao):
                count += 1
    return count

notebooks = contar_arquivos('.ipynb')
scripts_python = contar_arquivos('.py')
markdown_docs = contar_arquivos('.md')
csv_files = contar_arquivos('.csv')

print("=== Estatísticas do Projeto ===\n")
print(f"Notebooks Jupter:    {notebooks}")
print(f"Scripts Python:      {scripts_python}")
print(f"Documentação (MD):   {markdown_docs}")
print(f"Datasets (CSV):      {csv_files}")

# 2. Tamanho total do projeto (excluindo .venv)
total_size = 0
for root, dirs, files in os.walk(projeto_path):
    dirs[:] = [d for d in dirs if d not in ['.venv', '.git']]
    for file in files:
        filepath = os.path.join(root, file)
        total_size += os.path.getsize(filepath)

size_mb = total_size / (1024 * 1024)        
print(f"\nTamanho total:     {size_mb:.1f} MB")
```

```python
import subprocess

# Pega o histórico de commits
resultado = subprocess.run(
    ['git', 'log', '--oneline', '-20'],
    cwd=str(projeto_path),
    capture_output=True,
    text=True
)

print("=== Últimos 20 Commits ===\n")
print(resultado.stdout)

# Conta total de commits
resultado_total = subprocess.run(
    ['git', 'rev-list', '--count', 'HEAD'],
    cwd=str(projeto_path),
    capture_output=True,
    text=True
)

total_commits = resultado_total.stdout.strip()
print(f"Total de commits: {total_commits}")
```
